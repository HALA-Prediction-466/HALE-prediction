```python
import pandas as pd
from sklearn.preprocessing import StandardScaler, MinMaxScaler
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
# Load the cleaned dataset
file_path = '../data/cleaned_merged_dataset.csv'
data = pd.read_csv(file_path)
```

```python
# 1. Create cyclical features FIRST (before splitting)
data['Period_sin'] = np.sin(2 * np.pi * data['Period'] / data['Period'].max())
data['Period_cos'] = np.cos(2 * np.pi * data['Period'] / data['Period'].max())
```

```python
# 2. One-hot encoding (before splitting)
data = pd.get_dummies(
    data,
    columns=['ParentLocation', 'Location', 'Dim1'],
    drop_first=True
)
```

```python
# 3. Separate features and target
X = data.drop(columns=['HALE_60'])
y = data['HALE_60']
```

```python
train_data = data[data['Period'] <= 2014]
test_data = data[data['Period'] >= 2015]

X_train = train_data.drop(columns=['HALE_60'])
y_train = train_data['HALE_60']
X_test = test_data.drop(columns=['HALE_60'])
y_test = test_data['HALE_60']
```

```python
# 5. Define columns for different scaling methods
zscore_columns = [
   'HALE_Birth',
    'infant mortality rate (between birth and 11 months per 1000 live births)',
    'Age-standardized suicide rates (per 100 000 population)',
    'Alcohol, total per capita (15+ years) consumption (in litres of pure alcohol) (SDG Indicator 3.5.2)',
    'Estimates of rate of homicides (per 100 000 population)',
    'Mean Total Cholesterol (crude estimate)',
    'PM2.5_Exposure',
    'Health_Expenditure_pct_GDP'
]

percentage_columns = [
    'Prevalence of insufficient physical activity among adults aged 18+ years (age-standardized estimate) (%)',
    'Prevalence of obesity among adults, BMI ³ 30 (age-standardized estimate) (%)',
    'Prevalence of raised blood pressure among adults aged 30-79 years',
    'Probability of dying between the exact ages 30 and 70 years from cardiovascular diseases, cancer, diabetes, or chronic respiratory diseases (SDG 3.4.1)',
    'Undernourishment_Prevalence'
]
```

```python
# 6. Apply scaling on training set only
scaler_z = StandardScaler()
X_train[zscore_columns] = scaler_z.fit_transform(X_train[zscore_columns])
X_test[zscore_columns] = scaler_z.transform(X_test[zscore_columns])

scaler_mm = MinMaxScaler()
X_train[percentage_columns] = scaler_mm.fit_transform(X_train[percentage_columns])
X_test[percentage_columns] = scaler_mm.transform(X_test[percentage_columns])
```

```python
# 7. Save the processed datasets
X_train.to_csv('../data/X_train_normalized.csv', index=False)
X_test.to_csv('../data/X_test_normalized.csv', index=False)
y_train.to_csv('../data/y_train.csv', index=False)
y_test.to_csv('../data/y_test.csv', index=False)

print(f"Training set shape: {X_train.shape}")
print(f"Test set shape: {X_test.shape}")
print("Data preprocessing complete with proper train/test separation")
```
